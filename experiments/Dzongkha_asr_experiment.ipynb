# EXTRACT ZIP

```python
# ============================================
# EXTRACT ZIP
# ============================================

import zipfile
import os

zip_path = "/content/རྫོང་ཁ [dz] mp3.zip"

extract_path = "dzongkha_dataset"

with zipfile.ZipFile(
    zip_path,
    'r'
) as zip_ref:

    zip_ref.extractall(extract_path)

print("Extraction completed!")
```

# FIND MP3 FILES

```python
# ============================================
# FIND ALL MP3 FILES
# ============================================

mp3_files = []

for root, dirs, files in os.walk(extract_path):

    for file in files:

        if file.endswith(".mp3"):

            full_path = os.path.join(
                root,
                file
            )

            mp3_files.append(full_path)

print("TOTAL MP3 FILES:", len(mp3_files))

print("\nFIRST 5 FILES:\n")

for f in mp3_files[:5]:
    print(f)
```

# CONVERT MP3 → WAV

```python
# ============================================
# CONVERT MP3 TO WAV
# ============================================

from pydub import AudioSegment

wav_output_dir = "wav_files"

os.makedirs(
    wav_output_dir,
    exist_ok=True
)

for mp3_path in mp3_files:

    try:

        filename = os.path.basename(mp3_path)

        wav_name = filename.replace(
            ".mp3",
            ".wav"
        )

        wav_path = os.path.join(
            wav_output_dir,
            wav_name
        )

        # load mp3
        audio = AudioSegment.from_mp3(
            mp3_path
        )

        # ASR format
        audio = audio.set_frame_rate(16000)
        audio = audio.set_channels(1)

        # export wav
        audio.export(
            wav_path,
            format="wav"
        )

        print("Converted:", wav_name)

    except Exception as e:

        print("ERROR:", mp3_path)
        print(e)

print("\nALL CONVERSIONS COMPLETED!")
```

# LOAD DZONGKHA ASR MODEL

```python
# ============================================
# LOAD DZONGKHA ASR MODEL
# ============================================

from transformers import pipeline

asr = pipeline(
    "automatic-speech-recognition",
    model="etaidok/DzongkhaASR"
)

print("Dzongkha ASR model loaded!")
```

# TEST SINGLE DZONGKHA FILE

```python
# ============================================
# TEST SINGLE AUDIO FILE
# ============================================

test_file = os.path.join(
    wav_output_dir,
    os.listdir(wav_output_dir)[0]
)

print("Testing:")
print(test_file)

# ============================================
# RUN TRANSCRIPTION
# ============================================

result = asr(test_file)

print("\n====================")
print("DZONGKHA TRANSCRIPTION")
print("====================\n")

print(result["text"])
```

# TRANSCRIBE ALL FILES


```python
# ============================================
# BATCH TRANSCRIPTION (LIMITED FILES)
# ============================================

results = []

wav_files = os.listdir(wav_output_dir)

# take only first 5 files
wav_files = wav_files[:5]

print("TOTAL FILES USED:", len(wav_files))

for wav_file in wav_files:

    wav_path = os.path.join(
        wav_output_dir,
        wav_file
    )

    try:

        print("\nProcessing:", wav_file)

        result = asr(wav_path)

        text = result["text"]

        results.append({
            "file": wav_file,
            "transcription": text
        })

        print("\n====================")
        print("FILE:", wav_file)

        print("TEXT:")
        print(text)

    except Exception as e:

        print("\nERROR:", wav_file)
        print(e)
```

# SAVE TRANSCRIPTIONS

```python
# ============================================
# SAVE RESULTS
# ============================================

import pandas as pd

df = pd.DataFrame(results)

df.to_csv(
    "dzongkha_transcriptions.csv",
    index=False
)

print(df.head())
```

# LATENCY BENCHMARKING

```python
# ============================================
# LATENCY BENCHMARK
# ============================================

import time
import librosa

# ============================================
# AUDIO DURATION
# ============================================

audio, sr = librosa.load(
    test_file,
    sr=16000
)

audio_duration = len(audio) / sr

# ============================================
# PROCESSING TIME
# ============================================

start = time.time()

result = asr(test_file)

end = time.time()

processing_time = end - start

# ============================================
# REAL-TIME FACTOR
# ============================================

rtf = processing_time / audio_duration

print("\n====================")
print("LATENCY BENCHMARK")
print("====================")

print("Audio Duration:", audio_duration)

print("Processing Time:", processing_time)

print("Real-Time Factor:", rtf)
```
